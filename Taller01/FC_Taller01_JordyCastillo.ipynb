# **Ejercicio 2:**
Diseñe un programa en Fortran que calcule la altura máxima de un proyectil, conociendo el ángulo respecto a la horizontal y la rapidez inicial. Siga el proceso completo de la sección 2 y el estilo del ejemplo integrador de la sección 9 (diagrama de flujo, diccionario de datos, constantes con parameter , eco de entradas).

Fórmula: $h_{max} = \dfrac{v_0^2 \sin^2(\theta)}{2g}$, con $g = 9.8\ m/s^2$.

**Recuerde:** (sin) espera el ángulo en radianes, convierta antes de usarlo.

**Variables sugeridas:** v0 , angulo_grados , angulo_rad , g , h_max (todas
real ), y cuidado con no dividir enteros sin querer





### **SOLUCION:**

**Paso 1:** El problema nos pide hallar un valor maximo de la altura cuando es lanzado un proyectil teniendo en cuenta el angulo de lanzamiento. Segun la ecuacion para la altura maxima, entran valores de velocidad inicial del proyectil y su angulo de inclinacion, ambos siendo valores pertenecientes a los reales. Su unica restricción es la gravedad, la cual debe ser distinta de 0, pero en condiciones normales en la Tierra, la gravedad siempre se mantienen en 9.8m/s^2

**Paso 2:** Creamos el pseucodigo de la siguiente forma;
* Inicio
* Realizar comando %writefile para ejecutar el Fortran.
* ponerle un titulo al programa
* Definir constante (g) y el tipo de variable de las variables a usar.
* Dar la opción al usuario para digitalizar los valores que desee.
* Convertir grados a radianes
* Definir la ecuación a usar para las variables definidas.
* Redondear cifras significativas
* imprimir resultado
* Fin


```python
# Diagrama de flujo respectivo
from IPython.display import Image
Image(filename="/content/punto02.png")
```

```python
#PASO 3:

%%writefile AlturaDeProyectil.f90
program AlturaDeProyectil
    implicit none

!definimos variables con la clasificación (real)
    real :: v0 , angulo_grados , angulo_rad , hmax, hmax1    !variables a usar
    real, parameter :: g =9.8      !gravedad
    real, parameter :: pi=3.1416   !valor de Pi

!hacemos read para que el usuario digite sus numeros
print*, "Ingrese el valor de la Velocidad Inicial"
read*, v0

print*, "Ingrese el valor del Angulo de Lanzamiento"
read*, angulo_grados

!convertir grados a radianes
    angulo_rad = angulo_grados * pi/180

!realizamos formula

hmax= (v0**2 * (sin(angulo_rad)**2)) / (2*g)


!imprimir resultado
print*, "La altura maxima del proyectil es de: ", hmax, "metros"
print*, "Velocidad Inicial:", v0,  "Angulo de Lanzamiento:", angulo_grados, "grados"

end program AlturaDeProyectil
```

```python
#PASO 4:

!gfortran AlturaDeProyectil.f90 -o AlturaDeProyectil && ./AlturaDeProyectil
```

```python
#PASO 5:

%%writefile PASS_FAIL.f90
program PASS_FAIL
    implicit none

!definimos nuevas variables de prueba
    real :: v0_test , angulo_grados_test , angulo_rad_test , hmax01, hmax02, hmax03, diferencia
    real, parameter :: g = 9.8
    real, parameter :: pi=3.1416
    real, parameter :: tolerancia = 0.0001

!CASO 1:  90 grados,  9.8 m/s  , ESPERADO: 4.90 m
angulo_rad_test = 90 *(pi/180)
hmax01 = (9.8**2 * (sin(angulo_rad_test)**2)) / (2*g)
diferencia = abs(hmax01 - 4.90)

!hacemos un condicional para verificar si la diferencia es menor a la tolerancia y realizar el PASS o PASS_FAIL
if (diferencia < tolerancia) then
    print*, "CASO 1: PASS [Hmax:", hmax01,"m]"
else
    print*, "CASO 1: FAIL [Hmax:", hmax01,"m]"
end if


!CASO 2:  45 grados,  20 m/s  , ESPERADO: 10.2041 m
angulo_rad_test = 45 *(pi/180)
hmax02 = (20**2 * (sin(angulo_rad_test)**2)) / (2*g)
diferencia = abs(hmax02 - 10.2041)

if (diferencia < tolerancia) then
    print*, "CASO 2: PASS [Hmax:", hmax02,"m]"
else
    print*, "CASO 2: FAIL [Hmax:", hmax02,"m]"
end if


!CASO 3:  30 grados,  20 m/s  , ESPERADO: 5.1020 m
angulo_rad_test = 30 *(pi/180)
hmax03 = (20**2 * (sin(angulo_rad_test)**2)) / (2*g)
diferencia = abs(hmax03 - 5.1020)

if (diferencia < tolerancia) then
    print*, "CASO 3: PASS [Hmax:", hmax03,"m]"
else
    print*, "CASO 3: FAIL [Hmax:", hmax03,"m]"
end if



end program PASS_FAIL
```

# **Ejercicio 3:**
Glosario: defina con sus propias palabras: algoritmo, variable, constante, tipo de dato, diagrama de flujo, pseudocódigo, entrada de datos, salida de datos, implicit none, diccionario de datos, división entera, función intrínseca, operador lógico, error de sintaxis, error de ejecución, error lógico.






### **SOLUCION:**

**Algoritomo:** Es la secuencia de intrucciones logicas y especificas que resuelven un problema en cuestion.

<br>

**Variable:** Es un espacio o etiqueta dentro del sistema/lenguaje con un determinado nombre y valor, el cual puede ser manipulado durante el programa/codigo.

<br>

**Constante:** Es un valor fijo que no cambia durante la manipulacion del codigo, catalogado como "parameter" en Fortran

<br>

**Tipo de Dato:** Es la clasificación que tienen las variables de tal forma que el programa pueda diferenciarlas, tales como los numeros enteros, los reales, los caracteres o los logicos

<br>

**Diagrama de Flujo:** Es una representacion visual de los pasos de un algoritmo mediante un esquema con figuras geometricas

<br>

**Pseudocodigo:** Es la forma textual de describir a un algoritmo de forma que cualquier persona lo pueda entender, es decir, un lenguaje mas humano que de maquina.

<br>

**Entrada de Datos:** Son los datos y/o informacion que el programa recibe por parte del usuario, mediante el comando "read"

<br>

**Salida de Datos:** Son los datos y/o informacion que el programa entrega para ser vistos por el usuario, mediante el comando "print"

<br>

**Implicit none:** Es una intruccion fundamental de Fortran que cancela las declaraciones implicitas de variables, obligando a definir explicitamente el tipo de cada variable para prevenir futuros errores por parte del usuario.

<br>

**Diccionario de datos:** Son comentarios aclaratorios dentro del codigo por parte del usuario con el fin de aclarar las variables, sus unidades o demas caracteristicas fisicas de importancia.

<br>

**Division Entera:** Es una operacion entre enteros en la que el programa rechaza la parte decimal, entregando unicamente la parte entera.

<br>

**Funcion intriseca:** Son las funciones innatas del programa/lenguaje, como las raices, funciones trigonometricas o logaritmos.

<br>

**Operador Logico:** Son los condicionales matematicos que abarcan las condiciones logicas como AND, OR, NOT, estas devuelven booleanos.

<br>

**Error de Sintaxis** Es un fallo en el codigo por la gramatica utilizada u ortografia. El programa te avisa cuando algo no esta escrito de forma correcta, impidiendo realizar el codigo.

<br>

**Error de ejecucion:** Es un fallo en donde el programa detecta una operacion que no esta definida, deteniendo la ejecucion del codigo. Un ejemplo de ello es dividir entre cero o un logaritmo evaluado en cero.

<br>

**Error logico:** Es un fallo en el codigo que el programa no detecta, pero que genera inregularidades en los resultados. Estos errores solo los puede detectar el usuario al ver el resultado compilado.
